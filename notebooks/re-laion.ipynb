```python
import re

def generate_concept(dataset: list[str], concept: str) -> tuple[list[str], list[str]]:
    pattern = re.compile(f'(^|[\\s.,-:;]){concept}($|[\\s.,-:;])', flags=re.IGNORECASE)
    pos, neg = [], []
    for text in dataset:
        if text is None:
            continue
        if pattern.search(text) is not None:
            pos.append(text)
        else:
            neg.append(text)
    return pos, neg
```
